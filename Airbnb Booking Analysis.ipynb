```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
# Strip column names to remove any leading/trailing spaces
df.columns = df.columns.map(str).str.strip()

# Drop duplicate records
df.drop_duplicates(inplace=True)

# Drop 'house_rules' and 'license' safely
df.drop(columns=['house_rules', 'license'], inplace=True, errors='ignore')

# Fix potential typo in service fee column
if 'servce fee' in df.columns:
    df.rename(columns={'servce fee': 'service_fee_$'}, inplace=True)

# Handle price and service fee columns
for col in ['price', 'price_$', 'service fee', 'service_fee_$']:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)

# Rename columns properly
df.rename(columns={
    'price': 'price_$',
    'service fee': 'service_fee_$'
}, inplace=True)

# Convert to numeric safely
for col in ['price_$', 'service_fee_$']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop rows with missing values
df.dropna(inplace=True)

# Convert other columns to correct types
df['id'] = df['id'].astype(str)
df['host id'] = df['host id'].astype(str)
df['last review'] = pd.to_datetime(df['last review'], errors='coerce', dayfirst=True)
df['Construction year'] = df['Construction year'].astype(int)

# Correct spelling in neighbourhood group
df.loc[df['neighbourhood group'].str.lower() == 'brookln', 'neighbourhood group'] = 'Brooklyn'

# Remove outliers in 'availability 365'
if 'availability 365' in df.columns:
    df = df[df['availability 365'] <= 500]
```

```python
## 1. What are the different property types in the Dataset?
```

```python
property_types = df['room type'].value_counts().to_frame()
property_types
```

```python
room_type_bar = plt.bar(property_types.index,property_types.loc[:,"count"]);
plt.bar_label(room_type_bar,labels = property_types.loc[:,"count"],padding = 4);
plt.ylim([0,50000]);
plt.xlabel('Room Type');
plt.ylabel('Room Type Count');
plt.title('Property Types and their count in the Dataset');
```

```python
## 2.Which neighborhood group has the highest number of listings?
```

```python
hood_group = df['neighbourhood group'].value_counts().to_frame()
hood_group
```

```python
hood_group_bar = plt.bar(hood_group.index,hood_group.loc[:,"count"]);
plt.bar_label(hood_group_bar, labels = hood_group.loc[:,"count"],padding = 4);
plt.ylim([0,40000]);
plt.xlabel('Neighbourhood Groups');
plt.ylabel('Number of Listings');
```

```python
## 3. Which neighborhoods group have the highest average prices for Airbnb listings?
```

```python
avg_price = df.groupby('neighbourhood group')['price_$'].mean().sort_values(ascending= False).to_frame()
avg_price_bar = plt.bar(avg_price.index, avg_price.loc[:,'price_$']);
plt.bar_label(avg_price_bar,labels=round(avg_price.loc[:,"price_$"],2),label_type = 'edge',padding = 4);
plt.ylim([0,700]);
plt.xlabel('Neighbourhood Group');
plt.ylabel('Average Price per Listings ($)');
plt.xticks(rotation = 45);
plt.title('Average Price per Listings ($) in the Neighbourhood Group');
```

```python
## 4. Is there a relationship between the construction year of property and price?
```

```python
df.groupby(df['Construction year'])['price_$'].mean().to_frame().plot();
plt.xlabel('Construction Year');
plt.ylabel('Average Price ($)');
plt.title('Average Price ($) for Properties in each Construction Year');
```
